## Parámetro N = 100 + M

```python
NIA_LAST_TWO_DIGITS = [79, 95, 84, 54]
hm, M, N = th.compute_N(NIA_LAST_TWO_DIGITS)
k_mid = N // 2
print(f"Media armónica = {hm:.4f} -> M = {M}, N = {N}, k_mid = {k_mid}")
```

## Enumeración completa para N = 7 (operaciones bit a bit + generador)

```python
counts7 = ms.enumerate_counts(7)
samples = ms.sample_microstates(7)
for k in range(7, -1, -1):
    print(f"k={k}  tails={7-k}  {samples[k]}  Omega={counts7[k]}")
print("Total:", counts7.sum())
```

## Distribuciones de macroestados (log-gamma, sin overflow)

```python
fig, ax = plt.subplots(figsize=(7, 4))
for n in (20, 50, 80, 100):
    k, P = comb.distribution(n)
    ax.plot(k, P, "o-", ms=3, label=f"N={n}")
ax.set(xlabel="k (caras)", ylabel=r"$P_k$")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

## Probabilidades y entropía para N = 100 + M

```python
print(f"P(k=30)            = {float(comb.probability(N, 30)):.4e}")
print(f"P(k={k_mid})            = {float(comb.probability(N, k_mid)):.4e}")
print(f"P(N/2 ± 10)        = {comb.probability_range(N, k_mid - 10, k_mid + 10):.5f}")
print(f"P(N/2 ± 20)        = {comb.probability_range(N, k_mid - 20, k_mid + 20):.5f}")
for a, b in ((k_mid + 25, k_mid), (k_mid + 10, k_mid), (k_mid, 20)):
    x = comb.delta_S_over_kB(N, a, b)
    print(f"ΔS(k={a}, k={b}) / kB = {x:.4f}")
print(f"Espera para k=8    = {th.waiting_time_seconds(N, 8):.3e} s")
print(f"Cociente (i)       = {comb.likelihood_ratio(N, k_mid, M - 10, M + 10):.4f}")
```
